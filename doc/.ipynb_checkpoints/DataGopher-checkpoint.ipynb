# DataGopher Documentation 5-17-2026 #

DataGopher consists of four main applications, each one occupying a separate window and triggered by an appropriately named button: "Wrangle Data", "Pivot Data", "Plot Data", and  "Models".  Each button triggers  a python tkinter application located in a similarly named file.  The design philosophy is to provide the thinnest possible GUI layer between the python functions and the user.  This allows non-programmers to take advantage of python's tested, robust, scalable, and (best of all), FREE data science tools in a manageable open source package.

The "Get Data" button does what it says.  It opens a file chooser dialog box and allows you to choose from Excel .csv, .xlsx, and .xls files as well as some Stata .dta files (compatibility with stata data files depends on which Stata verison saved the file).   This feature could easily be extended to other file types.

The "Models" button makes various linear and generalized linear models available: OLS for linear regression, logit and probit for modeling binary response variables, negative binomial and Poisson which can be appropriate for positive integer response variables and gamma which is sometimes useful for positive continuous response variables.   These are but a few of the modelling options availble in python's statsmodels module, you should use the model that best fits your modelling task rather than choosing a model based on it's availabliity on a menu or the type of the response variable it expects.  "Models" also offers some of the typical plots for visualizing the model being fit. Models resides in the file goModel.py.

The "Plot Data" button, much as the name suggests, makes various plotting capabilities drawn from matplotlib and seaborn available (and resides in the file goPlot.py). 

The "Wrangle Data"  button offers variable transformation, row filtering, and variable renaming capabilities and is based on pandas and patsy (in the file goDataFilter.py).

The "Pivot Data" button offers  a spreadsheet style pivot table capability (and resides in the file goPivotb.py). 

The application keeps a running log of everything you've done.   If you want to preserve this log, the "Save Log" button is the button for you.  The default is to write it to a text file with a time stamped name that 
starts with "DG_Log"

The "Save Code" button saves a stand alone python script file that will reproduce grahics you have made with goPlot.py, models you have fit with goModel.py, and pivot tables you have created with pivotb.py.  Currently it will  also produce code to recapitulate filters, variable constructions, and renameings you have done with goDataFilter.py --but not reshapings. Saving the code for models and plots you have created allows you to customize your models and plots, and/or share them with others without having to laboriously reconstruct the sequence of keystrokes and menu choices you made in DataGopher.py.  *NOTE*: To run the code produced by DataGopher you will  need the python file graphics.py.  Just make sure that graphics.py is in the same folder (directory) as the generated python script.  

The right-most button, labeled "Graceful Exit", does exactly what it says.  It shuts down the DataGopher app and cleans up all remaining windows.

When you start Data Gopher you will see a small window with four buttons (eventually) pop onto the screen.   If you are running the application version of DataGopher this can take some time.  You can always recall a dismissed app.  The startup windows look something like this

![alt text](goStat-1a.png "Initial Window")

Let's click all the buttons. You may notice that nothing happens until you have chosen a data file.  So after opening a data file with the Get Data button, then click all the buttons.

![alt text](goStat-1b.png "Initial Window")

Admittedly, it's kind of a lot.  You may dismiss (red button) or hide (yellow button) all but the window you want to work with in order to minimize clutter.   You can always recall a dismissed/hidden app.  

Technical point: if you dismiss one of the windows and the click it's button to get it back later, the data will be syncrhonized automatically with the current data set.   If you minimize the window, should stay synchronized and the window may open a little quicker.    Why have two ways of de-cluttering?  When working with very large data sets it can take a little time to restart a window from scratch.   In that case minimizing the window may be more convenient than killing (dismissing) the window and re-creating it. With data sets under about 100K rows, it doesn't seem to make a difference. 

Everytime you choose a new data set, all of the open windows will resynchronize to the new data and the choices made in those windows will be reset to defaults.  The exception to this is the log file which keeps a running list of commands you have executed and parameter choices you have made.

## Data Wrangling ##

### How to use patsy notation ###

*This section uses some of the Python language to explain how Patsy notation works.   Patsy notation is used in DataGopher.  You do not need to know Python to use DataGopher but you will need to know a little Patsy notation to make full use of DataGopher.*

Patsy is a Python package that makes available a simple algebraic language for specifying statistical models (sometimes called experimental designs).  It is similar to the language used in R for specifying models, and to MatLab's formula languages for specifying models.   The language allows you to specify the dependent and independent variables, and to specify interactions.   As implemented in the Patsy package, the language allows for a wide variety of variable transformations as well.   It is the variable transformations we concentrate on here.  We will use this notation for specifying variable transformations, and for specifying models in the "Models" module.




Returning to our Diamonds.csv data, suppose we want to formulate a model to predict price based on color, cut, carat and depth.  Patsy notation for such a model would be: "price ~ color + cut + carat + depth". The "~" separates the dependent variable from the independent or explanatory variables. The "+" separates the indivdual explanatory variables. In this section we will only be interested in transforming variables, not in fully specifying the model hence we drop the "~" and anything to the left of it. 

Let's start by looking at the data:

```python
sb.scatterplot(df, x = 'carat', y= 'price',    s = 15)
```

The relationship doesn't look very linear, carat^2 or carat^3 might come closer to a linear relationship with price.  Let's look at what variables we have to work with.

```python
list(df.columns)
```

It would be nice if we had a variable that was carat cubed or carat squared.  Resist the urge to start writing some python code (it's hard, I know).  We can create a data set consisting of  price, carat, carat^2 and carat^3 using simple notation.   Our design formula will be price + carat + I(carat**2) + I(carat**3).   We'll use Patsy and the dmatrix function to do the work for us. 

```python
df2 = dmatrix('price + carat + I(carat**2) + I(carat**3)', data = df, return_type = 'dataframe', NA_action=NAAction(NA_types=[]))

df2.head(10)
```

Not bad for what amounts to one line of code!  The secret is in the I(...) that says to treat the contents of the parentheses as a set of arithmetic operations to be applied row-wise to the specified columns.  The other parameters in the dmatrix function call tell dmatrix to get its data from the dataframe df, to return a dataframe, and to put in a NA marker where ever it cannot complete the transformation either because of because of missing data or numerical failure (e.g. division by 0).   

Suppose that we want a variable that is the log of price

```python
df4 = dmatrix('np.log(price)', data = df, return_type = 'dataframe',NA_action=NAAction(NA_types=[]))
df4.head(10)
```

Note, with any numpy function (with prefix np.) we can skip the I().   We can also do logical operations

```python
df5 = dmatrix('price + I((price >= 300) & (price <= 400))', data = df, return_type = 'dataframe',NA_action=NAAction(NA_types=[]))
print(df5.iloc[25:35])
```

By now, a world of transformations may be occurring to you.  Here's a tricky one.   Suppose that we would like to create binary variables for each outcome of a the variable color.   Instead of a string variable with 7 entries 'D', 'E', ..., 'J' we would like 6 binary variables, one for each but one outcome.  Simply put a C() around the variable in question (be careful not to do this with variables that take on a lot of values).

```python
df6 = dmatrix('price + carat + C(color)', data = df, return_type = 'dataframe',NA_action=NAAction(NA_types=[]))
df6.head(10)
```

Trick question:  what happed to the variable for the outcome 'D'?

Answer: That's the "reference" outcome.   You can change the reference outcome by adding the parameter Treatment in the C(..) function. The line below makes 'J' the ommitted binary variable.

```python
df7 = dmatrix("price + carat + C(color,Treatment(reference='J'))", data = df, return_type = 'dataframe', NA_action=NAAction(NA_types=[]))
df6.head(10)
```

Suppose instead,  we wanted a binary variable that is 1 if color is either 'I' or 'J'. 

```python
df8 = dmatrix("I((color == 'I') | (color == 'J')) + color", data = df, return_type = 'dataframe', NA_action=NAAction(NA_types=[]))
df8.head(10)
```

For a somewhat more ambitious example, suppose we want to create a binary variable that is 1 if the price of that observation falls in the top 10 percent of all observed prices and 0 other wise.   We'll use the np.percentile() function, coupled with the logic operators.  

```python
df8 = dmatrix("I(price > np.percentile(price, 90)) + price", data = df, return_type = 'dataframe', NA_action=NAAction(NA_types=[]))
df8.head(10)
```

It's a bit difficult to see if this command really worked. The easiest way to exhibit the variable is to sum the created column.

```python
Nrow = sum(df8['Intercept'])
Nselectedrow = sum(df8['I(price > np.percentile(price, 90))[T.True]'])
print(f"Total number of rows: {Nrow} \nNumber of rows with price > 90th percentile: {Nselectedrow}")
```

### Missing Data: NaN or inf ###

Missing data is a pervasive but often overlooked problem.  A missing entry in a column is usually indicated by the "nan" (equivalently "NaN" or "na") string, which technically stands for "not a number" but also seems to indicate missing qualitative variables.  The natural approach is to simply remove rows with missing data.   For some calculations, for example computing regression coefficients, or plotting data this is essential.  You cannot do model fit calculations on data that does not exist,  you cannot plot data that does not exist.  If the missing data points are randomly distributed across the data set, this poses few problems.   On the other hand, when the occurance of missing data is correlated with one or more of the characteristics measured in the data set, simply removing all rows with missing data can yield very misleading results.  A classic example in the setting of a survey happens when a subject's likelihood of not responding to a question depends on the answer to that question.  For example, when questioned about income, individuals at the low end of the income distribution may be more reticent to respond than individuals above the median of the population. When questioned about alcohol use, the opposite may happen (those at the high end of the consumption distribution may be reticent to respond).  In this case, naively removing observations (rows) with missing data (indicating non-response to a survey question) may yield estimates of the population mean that are biased.  We can analyize patterns in the missing data by creating a variable that is 1 if the column contains a missing data point in a particular row and O otherwise.

Sometimes, the transformations we make to the data (for example square root or logarithm) will create missing entries in the new column when the calculated transformation cannot be done or yield infinite values for some outcomes of the data.  When Python thinks that the result of a calculation is either +infinity or -infinity it will return the string +inf or -inf (or +np.inf or -np.inf).  Most calculations with infinite values can yield unpredictable result.  Unfortunately, in a effort to be helpful,  these calculations are often flagged only as a warning and not as an error.   The result can be very peculiar results when fitting a model, and errors when trying to plot (infinity is very hard to depict in a finite plotting area).    

#### How DataGopher Deals with na and inf ###

The presence of "nan" in data used to create a plot or fit a model will generally cause a failure.  When doing operations on columns of data (for example variable transformations) the various Python commands are usually happy to skip over the entries containing na.  This may vary by the particular transformation being attempted.   When plotting or model fitting DataGopher will remove all rows containing na's.  Dealing with inf's is a more subtle problem as sometimes no indication of difficulty is given until a warning message is produced during a model fit calculation.  To make the process both clearer to the user and consistent, Datagopher will tolerate na's on input but will filter out rows with na values among the columns in the model or the columns to be plotted.  It will also report how many rows have been deleted (this is in the default plot title and is also posted in the model fit report as well as the log).  Datagopher will not tolerate inf's on input anywhere in the input data, or as a by-product  of data transformations.   On input the user will be told how many and in which columns the inf's occurred and will be offered the opportunity to convert inf's to na's.  The thinking here is that inf's on input are the product of infelicities in the data preparation upstream of DataGopher and should be rectified before proceeding, however, converting inf's into na's and continuing is a stop gap for those users don't have time to re-wrangle the data.  The rows with inf's in the relevant columns will be dropped before a model is fit or the data plotted (along with the other na's).  If the user does not need the columns with the inf's for plotting or modeling, then the resulting plots/models will have the same number of input rows as if the columns with the inf's had not been present, if the user must employ the columns with the inf's the user has been given fair warning that additional rows have been deleted.    

To illustrate the usefulness of analyzing missing data I've added two variables to the Diamonds data set.   A variable "diff_xy" which is the difference between "x" and "y" and a variable called "sqrt_diff_xy" which is the square root of the difference between "x" and "y".  You can see that there are lots of missing values in the "sqrt_diff_xy" column.  In this case it's easy to see that the occurance of missing data follows a suspicious pattern.  We can analyze the pattern by creating a new variable that is 1 if the entry in the "sqrt_diff_xy" is missing and 0 otherwise.  I'll call it "sqrt_diff_xy_missing".

![alt text](UC-6.png "Initial Window")

## Plot Data ##

Unsurprisingly, the Plot Data button brings up tools for plotting data.   The first thing you see is a simple window (below).  Which asks you to identify which variables you would like to plot.  


![alt text](PD-1.png "Initial Window")


 Python has a witheringly complex and detailed ecosystem of packages for graphing and plotting data.  Everything you might ever think about (and a few things you might not consider) with regards to representing data.  I collected only a few tools based on the matplotlib package (plotting in the sprirt of MatLab), and the Seaborn package.  Plot Data implements a rudimentary interface to plot 1, 2 and 3 dimensional representations of your data. 

Here is a brief outline of what's avaiable

* One Dimension
    * Histograms, with kernel density overlay
    * Box Plots 
    * Empirical Cumulative Distribution Function (ECDF)
    * Faceted grids of all (or most) of the above

* Two Dimension
    * 2D scatter plots or line plots
    * 2D scatter facet plots

* Three Dimension
    * 3D scatter plots

All plots start with some numerical quantity to be plotted.   Clicking on "Select Numerical Variables" will give a list of possibilities (non numeric variables excluded).  As you choose variables the list build from left to right.  One dimensional plots the leftmost variable controls the numerical values to be plotted.   For two dimensional scatter plots the first variable is the x-axis variable and the second variable the y-axis variable.  For three dimensions the first two variables are the independent variables (x and y) and the third variable will be the dependent variabe (z-axis).  You can add more dependent variables to the right.  Those variables can be plotted on top of the basic 2 or 3 dimensional scatter plot. The scale of the x axis (2D) or x and y axes (3D) are NOT adjusted as more variables are added to the plot.   Multiple scalse on the y (for 2D) and z (for 3D) axes are not allowed. You may have to fiddle with the axis limits to get a picture that makes sense or rescale a variable using the Wrangle Data module.

You can also pick variables to control the color, shape of marker or row and column facets.  These variables can be either numerical or non-numerical but cannot be one of the variables being plotted on an axis.  The variable chosen to be plotted, the color variable, marker variable, row facet variable and column facet variable must all be distinct.  Assigning one of them removes that choice from the available variables for the the others.  

There are some limitations.  You can use a variable to control the color of the markers and another variable to control the type of marker only when doing a 2D scatter plot.  If you want a  grid of plots where rows can correspond to one variable, the columns to the values of another variable and the color of the markers to yet a third variable.  You can do this by choosing 2D scatter Grid Plot. In this case if a variable is specified to control the marker type, it is ignored (three factors is a lot, four is too much).  3D plotting does not allow for a grid plot.  It will allow colors that depend on the value of a variable.

The button "Clear Last Variable" will do just that. The "Color Variable:" allows you to choose a variable to control the color of the dots in a scatter plot of the color of the lines in a line plot.  The plot type will give a list of available plot types.   Choosing a plot type will change the options available to you in the plot window.  The set of options are choices of which variables you wish to control the type of marker used for a data point in a scatter plot and the variables to be used to designate the rows and columns of a grid of plots.   For the marker and grid variables there is a limit on the number of outcomes the controlling variable may have.  This limit is driven by the practicality of plotting rather than any built-in limitation.  As you choose different plot types, you will be presented with a small number of options to decorate the plot these decorations also depend on the plot type.   Throughout all plots you will be able to choose the axis labels and a plot title.  If no title string is given the plot title will give the number of points plotted.  If you can't see all of a plot, try clicking the green button at the upper lefthand corner of the plot window, that will expand the window to full screen.  Tools for saving plots are embedded in the plot window.

### Scatter Plot Optons ###

You will note that for the 2D Scatter/Line plot and 3D Scatter plot options the variable window for dependent variables offers additional options.   Immeadiately below the marker size slider is a question asking if you want to add the variable to the scatter plot, after that you can choose to summarize the data with a lines (or surfaces) of different types and colors.   Choosing "-" as the line type results in no line being plotted.  

The problem when using line plots to represent scatter plot data is that in the most common setting there may be multiple values of the dependent variable associated with single values of the independent variable or variables. To create a line plot representation one needs to summarize these multiple values into a single values.   There are many ways to do this each one offering a different level of "smoothing". The options are listed under "Estimator/Smoother".  The current choices are

* <b>None:</b> this will attempt to draw the chosen line plot through each point.   Unless you know that there is only one value for the dependent variable for each indpedent variable value (as, for example, in time series data) this will lead to a fairly messy plot. 
* <b>mean:</b> chooses the mean of the dependent variable values associated with a single value of the independent variables, or a pair of values of the independent variables
* <b>median:</b> chooses the mean of the dependent variable values associated with a single value of the independent variables, or a pair of values of the independent variables
* <b>loess:</b> calculates a local estimated scatter plot smoothing of the dependent variable.  This is a popular "nonparemetric" apprach to smoothing data that fits a linear model at each point that takes a subset of close points of size equal to some fraction of the sample.  This fraction is set  in the "Loess Fraction" parameter.   The smaller the fraction the fewer points nearby points are sampled.  The algorithm used:  "LOESS_2D routine of Cappellari et al. (2013b), which implements
the multivariate LOESS algorithm of Cleveland & Devlin (1988)"
* <b>polynomial:</b> calculates a nth degree polynomial fit between the dependent and independent varible(s) and graphs it.   The order of the polynomial is controlled by the "Polynomial Order" parameter. 

Currently, confidence bounds are offered only for the polynomial smoothing.


## Models: Linear Models Tools ##

The Linear Models Tools window (hence forth "Models") allows you to choose a model to respresent your data.  The choices are ordinary least squares (OLS), logistic regression, probit, Poisson,  gamma and negative binomial.  Our data set is probably most amenable to least squares modeling.   Let's fit simple linear model to predict price as a function of carat. Click on the Dependent Variable drop down menu and select "price" for the Independent variable select carat. the model price ~ carat appears in the "Model:" text box.   Make sure that OLS is selected and then click Run Model.   You can see a regression report by clicking on the "Report" button.  You can see a log file of what the regression fit process did by clicking on "Log".  The specified models and reports are shown below.

**Note \#1:**. The model interface for Linear Models is pickier than the interface for the data wrangling module.  The dependent and independent variables must be chosen via the drop down menu.  Once chosen, one can type in any syntatically correct patsy expression using those variable names.  For example, if I choose "carat" as an independent variable, I can then type in I(carat\**2) + I(carat\**3).  To summarize, if the variable names are NOT chosen with the "Independent Variables:" drop down menu, the Models module will not "see" them.  But once it sees them you can modify them. 

**Note \#2:** The version of the models Model module used to make the graphics for this section was a little older than the current version of the code. ## Generalized Linear Models ##

### Logistic Regression: Logit ###

When the dependent variable is binary, some different modeling techniques come into play.  Linear Models offers two: logit and probit.  These maximum likelihood techniques are based on two different models of the process generating the data.  Earlier in these notes I created a binary variable which is 1 if the diamond in that observation was in the 75th percentile or above in sales price per carat and called it high_price_per_carat.  The purpose of my logit model will be to predict which diamonds in the sample fall into the 75% of price per carat using carat, color, cut, clarity, and table.  We can fit the logistic regression and plot the ROC curve (a visual indicator of the model's descriminatory power).  A common summary measure of a logit model's ability to predict the response variable (in this case high_price_per_carat) is AUC which for Area-Under-the-Curve.  The model achieved an astonishing 0.97507.   

![alt text](M-5.png "Initial Window")

This was a suspiciously strong result, but perhaps not surprising as carat was used as an independent variable and is also part of th definition of high_price_per_carat.  Let's look at some plots. ![alt text](M-6.png "Initial Window")

![alt text](M-7.png "Initial Window")

Note: Est. Mean Response means Estimated Mean Response.  The response variable is either 0 or 1 (either not above 75th percentile or above 75th percentile) so mean response = probability that diamond is at or above the 75th percentile in price per carat.  More sussinctly: for a logistic regression Est. Mean Response = estimate of P(esponse variable = 1).

### Other Generalized Linear Models ###

The remaining options under "Model Type" on the "Linear Models Tools" page are  Gamma, Probit, Poisson and Negative Binomial.  Each of these options refers to a different model.  Probit is another model that is useful for dependent variables with binary outcomes, Poisson and Negative Binomial are good for modeling data that consists of counts of the occurence of a particular event, and Gamma is good for non-negative continuous dependent variables.  Explaining each one in detail is beyond the scope of this documentation.  There are many very good references describing General Linear Model available on the Internet.

## Important But Somewhat Tedious Details ##

Original intent of DataGopher was to be a minimal set of, largely independent, component applications to support basic data wrangling and exploratory statistical modeling via a ``simple'' graphical user interface.  This is sometimes now called "unsupervised learning" in the sense that the user may be at the stage of sorting out what sort of model to use rather than testing a specific model formulation.  As such, an important design goal was to support a trial-and-error approach to graphics and modeling.   I have tried to keep annoying pop-up window advice to a minimum.  

Each of the buttons "Wrangle Data", "Pivot Data", "Plot Data" and "Models" corresponds to a separate tkinter class and each can be run as a stand alone app with minimal shared resources   If you look below the "if __name__ == "__main__":" line in each of the files corresponding to these four main functions ("goDataFilter.py", "goPivotb.py", "goPlot.py", "goModel.py" respectively), you will see the code that runs the corresponding app in standalone mode.  The files "Graphics.py", "Regression.py", "globalData.py", "TableDisplaya.py", "TableDisplayb.py", and "Listwidget.py"  contain the common resources needed by all of the four stand alone programs (in addition to a fairly standard set of imports). It is quite possible to run each of the four main applications separately simply invoke the corresponding python file instead of DataGopher.py.

In order to prevent incomplete or numerically invalid model fits, I have been somewhat finicky about what constitutes a successful model fit.  Specifically, many runtime warnings are treated as errors and will report a fit failure.  This is an attempt to avoid a situation where the model fit returns a report with lots of NA values.   It's difficult to have confidence in the parameter estimates when the model fit function returns coefficient values but at the same time returns NA for standard error.  This can happen when the model fit function runs into divide-by-0 conditions while fitting the model.  When DataGopher receives a runtime warning it will often stop and report an error.   It does not stop for language deprecation warnings, future code compatibilty warnings, etc.. (I hope).

### Pro Tips ###

Any tool that allows for rapid trial and error and model iteration runs the risk of the user finding a plot or a model they would like to reproduce but with little or no recollection of how they got to the model or plot.  This begs the question of how can one  double-check DataGopher?  How do we verify that the analysis is what we think it is, and how can we replicate that analysis outside fo DataGopher?  This is an issue that has gained greater salience as AI analytical assistance has gained popularity.   As of this writing, reproducability and verifiability are weaknesses in the AI assisted analytical workflow.   I mean this as no knock on AI tools, I have used them to produce this code.   But the purpose of AI is to learn from experience, and my data analysis task becomes part of my AI assistant's experience.  I should expect no more consistency than I would from a human co worker.  

DataGopher has some built in tools for verification and reproduction of results.  First of all, DataGopher keeps a date and time stamped log of everything it does.   Second, you can save the data that a model was fit on (e.g. after any rows with relevant missing data are removed), you can do the same for plots.  This feature is particularly useful if you have had to do some data wrangling to get the results you want.  What ever you may have done to the data ... or more politely ... however you may have curated the data ... you can save a copy of what data was used to fit the model or make a plot.  For models, you can also save a copy of the input data with the model predictions, residuals and confidence intervals.  

A second feature that is still under development is code generation.   DataGopher generates python code to run any model fit or plot from scratch using only standard libraries (numpy,pandas, matplotlib, statsmodels, seaborn etc.).   When you click on "Save Code" from the original button window the code for a plotting or model fitting session is saved in a text file .  The resulting python should be runnable in python without (much) editing.   As currently configured, the python script produced reads the original data in.   If you wish to apply the script to a different file you can change the input file path in the script.  If you have filtered the original data or created new variables, I suggest saving the changed data and then re-loading the changed data set before plotting and modelling.  The code generation procedure for plotting is also useful when you find that my choices of default formatting do not suit your needs.  Since the code is a mixture of plain matplotlib and seaborn, it is pretty easy to modify.   There is some boilerplate code at the top of the script file that imports the necessary python packages, and one include file called "Graphics.py" that contains some helper functions for plotting.  The file "Graphics.py" must be accessible to the generated code script, this is most easily accomplished by putting them in the same directory.  If you want to create a single stand-alone script, simply copy the functions in Graphics.py into the generated script file.

One important point that may be obvious to python programmers.  DataGopher depends on a collection of packages that must be installed in your python environment.  If a critical import is missing, DataGopher or the code that DataGopher produces will stop and experess it's discontent with the selection of imports on offer in your python environment.  If the last three sentences makes no sense to you, this might be a good time to learn about python environments, python package managers (conda, PIP) and import files.  Alternatively, there is an application version of DataGopher that bundles all of the needed resources together  (currently only available for Apple silicon a.k.a M-series) Macs.  This app is packaged by pyInstaller, requires about 350MB of disc space, and several minutes to start an it will generally fight with your Mac's security settings.  See my suggestions for running DataGophter below.

** Tip No. 1 ** If you want to do data wrangling before plotting and fitting models save the wrangled data and then read in the wrangled data for modeling and plotting.  That way you have a copy of the exact data used to produce the plots and fit the data.

** Tip No. 2 ** If you are impatient with tip No. 1 you can run the analysis and plotting on the wrangled data after saving it and then change the file path to the wrangled data in the generated script file.  This only works if you save the code. 

** Tip No. 3 ** For the truely impatient, when you wrangle data with goDataFilter.py, the data is automatically updated in the other modules so that you can move freely back and forth between modeling, plotting, pivot tables, and more data wrangling.  Each update will clear the information in the other modules' interfaces so be sure you remember the model you wanted to fit or plot you wanted to make.  While this approach is quicker and more convenient, you can lose track of things pretty easily.  I do this, I often regret it.

## How to Run DataGopher ##

##Download the Application (M-series Macs only)##

Download the folder "app" from the repository on GitHub from which this document originated.  Open the folder "app" and then "dist" and then double click on DataGopher.  This app is very slow to load, and will often conflict with your Mac's security settings.

### Jupyter Notebook or Jupyter Lab ###

Start Jupyter Notebook or Jupyter Lab, navigate to the 'srcnew directory' and open either the file "run_DataGopher_jupyter.ipynb" or "run_DataGopher_min.ipynb".  The cells in these two notebooks show you how to invoke DataGopher from Jupyter Notebook or Jupyter Lab and how to find the main storage object containing the variables used by DataGopher. 

### DIY ###

Download the contents of the "srcnew" directory to you local computer.   Navigate to your favorite Python development environment (Spyder, VSCode, IDLE3, PyCharm, emacs, etc.). Make sure that all of the necessary packages have been installed in your python implementation, fhen launch DataGopher.py.  Because DataGopher is a Tkinter app, it has it's own event loop that may conflict with your development environment.  You may have to fiddle with the matplotlib backend and your settings to get it to work conveniently.  Generally a simple development environment like IDLE is available with every Python distribution.


### Semi DIY (my favorite) ###

Use conda or miniconda to create an environment and install all of the necessary imports using conda and pip.  Once this environment is created, test it to make sure that it will run DataGophter.py.  Then create a shell script to launch the environment and DataGopher.     On Apple computers the shell script can be automagically turned into a clickable document and relocated to the dock. An example of this shell script for Mac and Linux and clickable document for Mac can be obtained from Dr. Knucklehead.  You can then launch you favorte python environment to do additional data wrangling, access database api's, and generally horse around with your data while using DataGopher for analysis and display.  This method also seems to result in the quickest startup on my Mac M4 machine.
