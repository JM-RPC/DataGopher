```python
import pandas as pd
import numpy as np
import statsmodels.api  as sm
import statsmodels.formula.api as smf
from statsmodels.genmod.generalized_linear_model import SET_USE_BIC_LLF
SET_USE_BIC_LLF(True) 
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import seaborn as sb
sb.set_style("darkgrid", {"grid.color": ".6", "grid.linestyle": ":"})
from mpl_toolkits.mplot3d import Axes3D
#import utilities2  ### some handy utilities for managing data to be used with statsmodels.
#from utilities2 import getData
#from loess.loess_2d import loess_2d #comment out if not available
```

```python
name =  ['Bob', 'Ted', 'Carol', 'Alice','Gerald','Pigeon']
```

```python
state = ['CA', 'CA', 'NV', 'AK', 'HI','HI']
```

```python
ascore = [1,2,3,4,5,6]
```

```python
bscore = [680, 720, 715, 770, 800, 680]
```

```python
cscore = [2.3, 3.3, 3.7, 3.3, 2.7, 3.5]
```

```python
gender = ['M', 'F','NB', 'M', 'F', 'NB']
```

```python
df = pd.DataFrame({'name':name, 'state':state, 'ascore':ascore, 'bscore':bscore, 'cscore':cscore, 'gender':gender})
```

```python
df.melt(id_vars = ['name'], value_vars = ['ascore', 'gender'], var_name = 'vname', value_name = 'valname')
```
