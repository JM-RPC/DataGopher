After quitting datagopher, internal data (input data, most recent model results, etc) are available in the "globals" dictionary.  Let's see what's inside of globals() by listing out the values of it's keys.

```python
globals().keys()
```

That's messy, lets get rid of the "system" variables (the ones that start with an underscore)

```python
#note flashy python code
", ".join([item for item in list(globals().keys()) if not item.startswith('_')])
```

 'gdata' is an instance of the class globalData() (in globalData.py) in which datagopher stores data shared among the various modules of DataGopher.  'imdl' is and instance of the class modelData() (in Regression.py) which contains the internal data for the most recent model fit.  globals()['gdata'] accesses a member of that object.  For example the current data set is in globals()['gdata'].data  the current model

```python
globals()['gdata'].data.head(5)
```

An easy thing to do is to fetch the gdata and imdl classes and rename them you can manipulate them more easily.

```python
myobj = globals()['gdata']
mymodel = globals()['imdl']
```

```python
print("_________________myobj (gdata)_______________________________________")
print(", ".join([item for item in  list(dir(myobj)) if not item.startswith('_')]))
print("=====================================================================")
print("_________________mymodel (imdl)______________________________________")
    
print(", ".join([item for item in  list(dir(mymodel)) if not item.startswith('_')]))
print("=====================================================================")
```

The easiest way to see the complete inventory of what you can get from gdata is to look at the class specification globalData.py.  Handily reproduced below:


self.data  #----> the input data
self.fpath  #----> path to the input data
self.model_res  #----> last model results
self.model_data  #----> copy of the data used to fit the last model
self.datatable #----> the last pivot table produced
self.logstr = ''       #----> the log (a long string)
self.codestr = ''       #----> code produced in this session (a long string)
